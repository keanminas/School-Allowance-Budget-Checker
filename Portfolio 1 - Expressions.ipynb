## How much money do I have left after school?

## Objective
The goal of this program is to calculate a student's remaining daily allowance by subtracting various expenses (food, fare, and others) from a starting budget. This demonstrates the use of variable assignments and arithmetic expressions in a real-life financial planning scenario.

```python
# Fixed Budget for school day
money = 150
food = 60
fare = 30

# Additional expenses
others = 50 # Change this value depending on what other expenses you made

# Total expenses for the day
total = (food + fare + others)
remaining = money - (food + fare + others)

# Output results
print("starting money:", money)
print("total expenses:", total)
print("remaining money :", remaining)
```
