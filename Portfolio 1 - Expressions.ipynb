## How much money do I have left after school?

## Objective
The goal of this program is to calculate a student's remaining daily allowance by subtracting various expenses (food, fare, and others) from a starting budget. This demonstrates the use of variable assignments and arithmetic expressions in a real-life budgeting.

```python
# Setting up the initial budget
money = 150

# Listing down the daily expenses
food = 60
fare = 30
others = 40 # Change this value depending on the additional expenses made

# Calculating the sum of all the expenses
total = (food + fare + others)

# Subtracting the total expenses from the starting money
remaining = money - total

# Showing the final results
print("starting money:", money)
print("total expenses:", total)
print("remaining money :", remaining)
```
